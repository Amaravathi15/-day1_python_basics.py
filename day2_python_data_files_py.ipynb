```python
# 1. List Comprehension - Extract Even Numbers

def extract_even_numbers(numbers):
    return [num for num in numbers if num % 2 == 0]


print("Even Numbers Example 1:", extract_even_numbers([1, 2, 3, 4, 5, 6]))
print("Even Numbers Example 2:", extract_even_numbers([10, 15, 20, 25]))
```

```python
# 2. Dictionary Comprehension - Number Squares
# ----------------------------------------------------------

def number_squares(numbers):
    return {num: num ** 2 for num in numbers}


print("Squares Example 1:", number_squares([1, 8, 12]))
print("Squares Example 2:", number_squares([4, 5, 6]))
```

```python
# 3. Lambda - Sort Tuples by Second Value
# ----------------------------------------------------------

def sort_by_second_value(data):
    return sorted(data, key=lambda x: x[1])


print("Sort Example 1:",
      sort_by_second_value([("a", 3), ("b", 1), ("c", 2)]))

print("Sort Example 2:",
      sort_by_second_value([("x", 10), ("y", 5), ("z", 8)]))
```

```python
# 4. map() - Convert Strings to Uppercase
# ----------------------------------------------------------

def convert_to_uppercase(words):
    return list(map(str.upper, words))


print("Uppercase Example 1:",
      convert_to_uppercase(["apple", "banana"]))

print("Uppercase Example 2:",
      convert_to_uppercase(["python", "data"]))
```

```python
# 5. filter() - Words Longer Than 4 Characters
# ----------------------------------------------------------

def filter_long_words(words):
    return list(filter(lambda word: len(word) > 4, words))


print("Filter Example 1:",
      filter_long_words(["cat", "elephant", "dog", "tiger"]))

print("Filter Example 2:",
      filter_long_words(["data", "science", "ai", "machine"]))
```

```python
#6. Read TXT File and Count Lines & Words
# ----------------------------------------------------------

def count_lines_and_words(file_name):
    line_count = 0
    word_count = 0

    with open(file_name, "r") as file:
        for line in file:
            line_count += 1
            word_count += len(line.split())

    return line_count, word_count

    # Example usage:
# lines, words = count_lines_and_words("sample.txt")
# print("Lines:", lines)
# print("Words:", words)
```

```python
#7. Try/Except - Divide by Zero & Invalid Input
# ----------------------------------------------------------

def safe_division():
    try:
        num1 = float(input("Enter first number: "))
        num2 = float(input("Enter second number: "))

        result = num1 / num2
        print("Result:", result)

    except ZeroDivisionError:
        print("Error: Cannot divide by zero.")

    except ValueError:
        print("Error: Invalid input. Enter numeric values.")
# Example usage:
# safe_division()
```

```python
# 8. Read CSV and Print Rows Based on Condition
# ----------------------------------------------------------

def filter_csv_rows(file_name, column_index, threshold):
    """
    Print rows where selected column value > threshold
    """

    with open(file_name, "r") as file:
        header = file.readline()
        print("Matching Rows:")

        for line in file:
            row = line.strip().split(",")

            try:
                if float(row[column_index]) > threshold:
                    print(row)

            except (ValueError, IndexError):
                continue
     # Example usage:
# filter_csv_rows("sales.csv", 2, 1000)
```
