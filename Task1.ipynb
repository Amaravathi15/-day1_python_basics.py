```python
##2. Factorial of a number

def factorial(n):
    result = 1

    for i in range(1, n + 1):
        result *= i

    return result

print(factorial(5))
```

```python
##3.Prime number check

def is_prime(num):
    if num <= 1:
        return False

    for i in range(2, num):
        if num % i == 0:
            return False

    return True


print("\nPrime Tests:")
print(is_prime(11))
print(is_prime(12))
```

```python
 ##4.Reverse a string (no built-in reverse)

 def reverse_string(text):
    reversed_text = ""

    for char in text:
        reversed_text = char + reversed_text

    return reversed_text


print("\nReverse String Tests:")
print(reverse_string("Amaravathi"))
print(reverse_string("Joy"))

```

```python
#5. Palindrome check
def palindrome(text):
    reverse = ""

    for ch in text:
        reverse = ch + reverse

    if text == reverse:
        return True
    else:
        return False

print(palindrome("madam"))
print(palindrome("hello"))
```

```python
#6. Word count of a sentence

def word_count(sentence):
    words = sentence.split()
    return len(words)

print(word_count("I love Python"))
print(word_count("Python is a easy language to learn"))
```

```python
# 7. Character Frequency of a Word

def char_frequency(word):
    freq = {}

    for char in word:
        if char in freq:
            freq[char] = 1
        else:
            freq[char] = 1

    return freq

print(char_frequency("Goavardhan"))
print(char_frequency("Gautham"))
```

```python
# 8. Remove Duplicates from a List (Keep Order)

def remove_duplicates(lst):
    unique = []

    for item in lst:
        if item not in unique:
            unique.append(item)

    return unique
print(remove_duplicates([1, 2, 2, 3, 4, 3]))
print(remove_duplicates(["a", "b", "a", "c"]))
```

```python
# 9. Find Max and Min in a List (Without max/min)

def find_max_min(numbers):
    maximum = numbers[0]
    minimum = numbers[0]

    for num in numbers:
        if num > maximum:
            maximum = num

        if num < minimum:
            minimum = num

    return maximum, minimum

print(find_max_min([10, 5, 20, 8, 15]))

```

```python
#10. Sum of even numbers from 1 to N

def sum_even_numbers(n):
    total = 0

    for i in range(1, n + 1):
        if i % 2 == 0:
            total += i

    return total

print(sum_even_numbers(10))
print(sum_even_numbers(20))
```

```python
#11. Build a dictionary from two lists (keys, values)

def build_dictionary(keys, values):
    result = {}

    for i in range(len(keys)):
        result[keys[i]] = values[i]

    return result


print(build_dictionary(["a", "b", "c"], [1, 2, 3]))
print(build_dictionary(["name", "age"], ["John", 25]))

```

```python
# 12. Count Vowels in a String

def count_vowels(text):
    vowels = "aeiouAEIOU"
    count = 0

    for char in text:
        if char in vowels:
            count += 1

    return count

print(count_vowels("Python"))
print(count_vowels("Education"))
```

```python
#14. Grade calculator (score 0-100 -> letter grade)


def calculate_grade(score):
    if score >= 90:
        return "A"
    elif score >= 80:
        return "B"
    elif score >= 70:
        return "C"
    elif score >= 60:
        return "D"
    else:
        return "F"


print(calculate_grade(95))
print(calculate_grade(72))
```

```python
#15. Simple calculator (two numbers + operator, handle divide-by-zero)


def calculator(num1, num2, operator):

    if operator == "+":
        return num1 + num2

    elif operator == "-":
        return num1 - num2

    elif operator == "*":
        return num1 * num2

    elif operator == "/":
        if num2 == 0:
            return "Error: Division by zero"
        return num1 / num2

    else:
        return "Invalid Operator"

print(calculator(10, 5, "+"))
print(calculator(10, 5, "/"))
print(calculator(10, 0, "/"))
```
