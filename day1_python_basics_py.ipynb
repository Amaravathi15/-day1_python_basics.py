```python
def fizz_buzz():
    for i in range(1, 51):
        if i % 3 == 0 and i % 5 == 0:
            print("FizzBuzz")
        elif i % 3 == 0:
            print("Fizz")
        elif i % 5 == 0:
            print("Buzz")
        else:
            print(i)

print("1. Fizz Buzz")
fizz_buzz()
```

```python
# 2. Factorial of a Number
# --------------------------------------------------
def factorial(n):
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result


print("Factorial Tests:")
print(factorial(6))
print(factorial(7))
```

```python
# 3. Prime Number Check
# --------------------------------------------------
def is_prime(n):
    if n <= 1:
        return False

    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False

    return True


print("\nPrime Tests:")
print(is_prime(11))
print(is_prime(12))
```

```python
# 4. Reverse a String (No built-in reverse)
# --------------------------------------------------
def reverse_string(text):
    reversed_text = ""

    for char in text:
        reversed_text = char + reversed_text

    return reversed_text


print("\nReverse String Tests:")
print(reverse_string("Amaravathi"))
print(reverse_string("Joy"))
```

```python
# 5. Palindrome Check
# --------------------------------------------------
def is_palindrome(text):
    return text == reverse_string(text)


print("\nPalindrome Tests:")
print(is_palindrome("madam"))
print(is_palindrome("python"))
```

```python
# 6. Word Count of a Sentence
# --------------------------------------------------
def word_count(sentence):
    words = sentence.split()
    return len(words)


print("\nWord Count Tests:")
print(word_count("I love Python"))
print(word_count("Python is a easy language to learn"))
```

```python
# 7. Character Frequency of a Word
# --------------------------------------------------
def char_frequency(word):
    freq = {}

    for char in word:
        if char in freq:
            freq[char] += 1
        else:
            freq[char] = 1

    return freq


print("\nCharacter Frequency Tests:")
print(char_frequency("Goavardhan"))
print(char_frequency("Gautham"))
```

```python
# 8. Remove Duplicates from a List (Keep Order)
# --------------------------------------------------
def remove_duplicates(lst):
    unique = []

    for item in lst:
        if item not in unique:
            unique.append(item)

    return unique


print("\nRemove Duplicates Tests:")
print(remove_duplicates([1, 2, 2, 3, 4, 3]))
print(remove_duplicates(["a", "b", "a", "c"]))
```

```python
# 9. Find Max and Min in a List (Without max/min)
# --------------------------------------------------
def find_max_min(lst):
    maximum = lst[0]
    minimum = lst[0]

    for num in lst:
        if num > maximum:
            maximum = num

        if num < minimum:
            minimum = num

    return maximum, minimum


print("\nMax-Min Tests:")
print(find_max_min([10, 5, 30, 2, 50]))
print(find_max_min([7, 7, 1, 9]))
```

```python
# 10. Sum of Even Numbers from 1 to N
# --------------------------------------------------
def sum_even_numbers(n):
    total = 0

    for i in range(1, n + 1):
        if i % 2 == 0:
            total += i

    return total


print("\nSum of Even Numbers Tests:")
print(sum_even_numbers(10))
print(sum_even_numbers(20))
```

```python
# 11. Build Dictionary from Two Lists
# --------------------------------------------------
def build_dictionary(keys, values):
    result = {}

    for i in range(len(keys)):
        result[keys[i]] = values[i]

    return result


print("\nDictionary Tests:")
print(build_dictionary(["a", "b", "c"], [1, 2, 3]))
print(build_dictionary(["name", "age"], ["John", 25]))
```

```python
# 12. Count Vowels in a String
# --------------------------------------------------
def count_vowels(text):
    vowels = "aeiouAEIOU"
    count = 0

    for char in text:
        if char in vowels:
            count += 1

    return count


print("\nVowel Count Tests:")
print(count_vowels("Python"))
print(count_vowels("Education"))
```

```python
# 13. Celsius <-> Fahrenheit Converter
# --------------------------------------------------
def celsius_to_fahrenheit(celsius):
    return (celsius * 9 / 5) + 32


def fahrenheit_to_celsius(fahrenheit):
    return (fahrenheit - 32) * 5 / 9


print("\nTemperature Conversion Tests:")
print(celsius_to_fahrenheit(0))
print(celsius_to_fahrenheit(100))
print(fahrenheit_to_celsius(32))
print(fahrenheit_to_celsius(212))
```

```python
# 14. Grade Calculator
# --------------------------------------------------
def calculate_grade(score):
    if score >= 90:
        return "A"
    elif score >= 80:
        return "B"
    elif score >= 70:
        return "C"
    elif score >= 60:
        return "D"
    else:
        return "F"


print("\nGrade Calculator Tests:")
print(calculate_grade(95))
print(calculate_grade(72))
```

```python
# 15. Simple Calculator
# --------------------------------------------------
def calculator(num1, num2, operator):

    if operator == "+":
        return num1 + num2

    elif operator == "-":
        return num1 - num2

    elif operator == "*":
        return num1 * num2

    elif operator == "/":
        if num2 == 0:
            return "Error: Division by zero"
        return num1 / num2

    else:
        return "Invalid Operator"


print("\nCalculator Tests:")
print(calculator(10, 5, "+"))
print(calculator(10, 5, "/"))
print(calculator(10, 0, "/"))
```
